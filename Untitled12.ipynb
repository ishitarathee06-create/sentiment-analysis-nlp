```python
import pandas as pd, zipfile, glob, os

# 1) find the uploaded zip anywhere in Colab's files
zips = glob.glob("/content/**/*.zip", recursive=True)
print("Zip files found:", zips)

# 2) unzip it
for z in zips:
    with zipfile.ZipFile(z) as f:
        f.extractall("/content/data")

# 3) find Tweets.csv inside
csvs = glob.glob("/content/data/**/Tweets.csv", recursive=True)
print("CSV found:", csvs)

df = pd.read_csv(csvs[0])[["text", "airline_sentiment"]]
df.columns = ["text", "sentiment"]

print()
print("Rows and columns:", df.shape)
print()
print(df["sentiment"].value_counts())
print()
print("Empty values:", df.isnull().sum().sum())
df.head()
```

```python
import re, nltk
nltk.download("stopwords")
from nltk.corpus import stopwords

KEEP = {"not", "no", "nor", "never", "very", "too", "but", "against", "only"}
STOP = set(stopwords.words("english")) - KEEP

CONTRACTIONS = {
    "won't": "will not", "can't": "can not", "cannot": "can not", "ain't": "is not",
    "wont": "will not", "cant": "can not", "dont": "do not", "doesnt": "does not",
    "didnt": "did not", "isnt": "is not", "wasnt": "was not", "arent": "are not",
    "werent": "were not", "havent": "have not", "hasnt": "has not", "hadnt": "had not",
    "couldnt": "could not", "wouldnt": "would not", "shouldnt": "should not",
    "im": "i am", "ive": "i have", "youre": "you are",
}
EMOJI = {"😊": " happy ", "😀": " happy ", "😃": " happy ", "😄": " happy ", "🙂": " happy ",
         "😍": " love ", "❤": " love ", "👍": " good ",
         "😡": " angry ", "😠": " angry ", "😞": " sad ", "😢": " sad ", "😭": " sad ",
         "🙁": " sad ", "😔": " sad ", "👎": " bad "}

def clean_text(text):
    t = str(text).lower().replace("’", "'").replace("‘", "'")
    for e, w in EMOJI.items():
        t = t.replace(e, w)
    t = re.sub(r"http\S+|www\.\S+", " ", t)
    t = re.sub(r"@\w+", " ", t)
    t = t.replace("#", " ")
    t = re.sub(r"[^a-z'\s]", " ", t)
    t = " ".join(CONTRACTIONS.get(w, w) for w in t.split())
    t = re.sub(r"n't\b", " not", t)
    t = re.sub(r"'m\b", " am", t)
    t = re.sub(r"'re\b", " are", t)
    t = re.sub(r"'ve\b", " have", t)
    t = re.sub(r"'ll\b", " will", t)
    t = re.sub(r"'d\b", " would", t)
    t = re.sub(r"'s\b", "", t)
    t = re.sub(r"[^a-z\s]", " ", t)
    words = [w for w in t.split() if w not in STOP]
    return " ".join(words)

for s in ["I'm not happy!", "I don't like this", "im sad", "I am happy 😊",
          "@VirginAmerica it's really aggressive to blast... http://t.co/xyz"]:
    print(repr(s), "->", repr(clean_text(s)))
```

```python
df["clean"] = df["text"].apply(clean_text)

# remove tweets that became empty after cleaning (e.g. only a link or @mention)
before = len(df)
df = df[df["clean"].str.strip() != ""].reset_index(drop=True)
print("Tweets before:", before)
print("Tweets after removing empty ones:", len(df))
print()
print(df[["text", "clean", "sentiment"]].head(8).to_string())
```

```python
import pandas as pd

pos_w = ["happy","glad","excited","great","good","amazing","awesome","wonderful","fantastic",
         "excellent","satisfied","pleased","delighted","grateful","brilliant","perfect","lovely",
         "nice","impressed","thrilled","cheerful","proud","fine","well"]
neg_w = ["sad","unhappy","angry","upset","bad","terrible","horrible","awful","disappointed",
         "disappointing","frustrated","annoyed","poor","useless","disgusting","miserable",
         "furious","unsatisfied","worried","rude"]

feel_t = ["i am {}","i am so {}","i am really {}","i am feeling {}","i feel {}","this is {}",
          "it was {}","the service was {}","{}","very {}","so {}"]
neg_t  = ["i am not {}","i am not so {}","i am not really {}","i am not feeling {}","i do not feel {}",
          "i don't feel {}","i'm not {}","im not {}","this is not {}","it was not {}","it is not {}",
          "not {}","the service was not {}","not very {}"]
soft_t = ["this is not {}","it was not {}","it is not {}","not {}","the service was not {}"]

rows = []
for w in pos_w:
    rows += [(t.format(w), "positive") for t in feel_t]
    rows += [(t.format(w), "negative") for t in neg_t]
for w in neg_w:
    rows += [(t.format(w), "negative") for t in feel_t]
for w in ["bad","terrible","horrible","awful","poor"]:
    rows += [(t.format(w), "positive") for t in soft_t]

pos_s = ["i love this","i like this","i love it","i like it","i really like this","i enjoyed it",
 "i would recommend this","i would recommend it","highly recommended","thank you so much","thanks a lot",
 "best experience ever","this is the best","everything was perfect","i am satisfied",
 "i am very happy with the service","great job","well done","good service","very good service",
 "excellent service","nice work","i am impressed","works perfectly"]
neg_s = ["i hate this","i hate it","i dislike this","i do not like this","i don't like this","i dont like it",
 "i do not like it","i do not love this","i would not recommend this","i don't recommend this","never again",
 "worst experience ever","this is the worst","everything was terrible","i am not satisfied",
 "i am not happy with the service","bad service","very bad service","terrible service","not good",
 "not great","no good","not worth it","waste of money","waste of time","i am disappointed",
 "very disappointing","does not work","not working","i did not enjoy it","i didn't enjoy this",
 "i am not impressed","not recommended"]
neu_s = ["the meeting is at five pm","i am going to the market","today is monday","the shop opens at nine",
 "i have a flight tomorrow","it is a table","i am reading a book","the report is on the desk",
 "the train arrives at noon","i live in a small town","this is a pen","the class starts at ten",
 "i will call you later","the office is on the second floor","i went to the store","the file is attached",
 "please send the document","what time does the flight leave","where is the bus stop",
 "how do i reset my password","it is cloudy today","i am at home","we are meeting tomorrow",
 "the project is due on friday","the package arrived today","i am waiting for the bus",
 "the website has three pages","this is a laptop","the book is on the shelf","i usually wake up at seven",
 "the store is near the station","my flight is at noon","i booked a ticket","the movie is two hours long",
 "he is a teacher","she is my sister","the event is on saturday","i work in the city",
 "the price is five hundred rupees","i am learning python","i have a class today",
 "the room is on the first floor","the car is parked outside","i have an appointment on monday"]
rows += [(s,"positive") for s in pos_s] + [(s,"negative") for s in neg_s] + [(s,"neutral") for s in neu_s]

aug = pd.DataFrame(rows, columns=["text","sentiment"])
aug = pd.concat([aug]*3, ignore_index=True)
aug["clean"] = aug["text"].apply(clean_text)

print("Extra sentences:", len(aug))
print()
print(aug["sentiment"].value_counts())
print()
print(aug.sample(8, random_state=1)[["text","clean","sentiment"]].to_string())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Split ONLY the real tweets, so the test score is honest
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df["sentiment"])

# Training data = 80% of real tweets + the extra sentences
train_all = pd.concat([train_df[["clean", "sentiment"]], aug[["clean", "sentiment"]]], ignore_index=True)

# TF-IDF with word pairs and triples, so "not happy" is learned as one feature
vectorizer = TfidfVectorizer(ngram_range=(1, 3), min_df=2, sublinear_tf=True)
X_train = vectorizer.fit_transform(train_all["clean"])

model = LogisticRegression(max_iter=2000, C=5, class_weight="balanced")
model.fit(X_train, train_all["sentiment"])

# Test on unseen real tweets
pred = model.predict(vectorizer.transform(test_df["clean"]))
print("Training rows:", len(train_all), "| Test rows:", len(test_df))
print("Accuracy:", round(accuracy_score(test_df["sentiment"], pred), 4))
print()
print(classification_report(test_df["sentiment"], pred))

labels = ["negative", "neutral", "positive"]
cm = confusion_matrix(test_df["sentiment"], pred, labels=labels)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=labels, yticklabels=labels)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
def predict_sentiment(text):
    if not text or not text.strip():
        return {"Please enter some text": 1.0}
    cleaned = clean_text(text)
    if cleaned.strip() == "":
        return {"neutral": 1.0}
    X = vectorizer.transform([cleaned])
    probs = model.predict_proba(X)[0]
    return {cls: float(p) for cls, p in zip(model.classes_, probs)}

# quick tests
for s in ["I am not happy", "great service, thank you", "the flight is at noon", "worst experience ever"]:
    print(s, "->", predict_sentiment(s))
```

```python
def predict_sentiment(text):

    cleaned = clean_text(text)

    if cleaned.strip() == "":
        return "Invalid Input"

    text_vector = vectorizer.transform([cleaned])

    prediction = model.predict(text_vector)[0]

    probabilities = model.predict_proba(text_vector)[0]

    confidence = max(probabilities) * 100

    return prediction, round(confidence, 2)
```

```python
test_sentences = [
    "I absolutely love this service!",
    "This is the worst experience ever.",
    "The flight is scheduled for 5 PM.",
    "I am not happy with the service.",
    "Everything was wonderful!"
]

for sentence in test_sentences:

    result, confidence = predict_sentiment(sentence)

    print("Text:", sentence)
    print("Sentiment:", result)
    print("Confidence:", confidence, "%")
    print("-" * 50)
```
